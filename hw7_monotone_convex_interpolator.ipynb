### IFR curve plot and interpolation choice

The interpolator uses a cubic Hermite segment between each pair of nodes. For a rising segment with secant slope $d$, it uses endpoint slopes $d-c$ and $d+c$, with $c$ limited by the neighboring secant changes and by $d/4$. This keeps the segment increasing and convex while passing through both supplied nodes. If the node pattern does not support a rising convex segment, that interval is linear. A one-node curve is constant. The curve is continuous at every node; the slope can change at a node.

```python
# Plot the supplied IFR nodes and the implemented interpolators.
import matplotlib.pyplot as plt
import numpy as np
from fixedincomelib.utilities import Interpolator1DMonotoneConvex, ExtrapMethod

value_date = pd.Timestamp("2026-01-02")
node_times = np.array([(pd.Timestamp(d) - value_date).days / 365.0 for d in node_dates])
plot_times = np.linspace(0.0, node_times[-1], 500)

sofr_interpolator = Interpolator1DMonotoneConvex(
    node_times, sofr_quotes["values"].to_numpy(), ExtrapMethod.FLAT
)
ff_interpolator = Interpolator1DMonotoneConvex(
    node_times, ff_quotes["values"].to_numpy(), ExtrapMethod.FLAT
)
sofr_curve = [sofr_interpolator.interpolate(t) for t in plot_times]
ff_curve = [ff_interpolator.interpolate(t) for t in plot_times]

plt.figure(figsize=(9, 5))
plt.plot(plot_times, sofr_curve, label="SOFR IFR", color="navy")
plt.plot(plot_times, ff_curve, label="FF IFR", color="darkorange")
plt.scatter(node_times, sofr_quotes["values"], color="navy", marker="o")
plt.scatter(node_times, ff_quotes["values"], color="darkorange", marker="o")
plt.xlabel("Time from value date (years)")
plt.ylabel("Instantaneous forward rate")
plt.title("Monotone-Convex Interpolation of Supplied IFR Nodes")
plt.grid(alpha=0.25)
plt.legend()
plt.tight_layout()
plt.show()
```

The products have different cash-flow dates and depend on different projected rates, so the interpolation change affects their PVs by different amounts and can have either sign.
